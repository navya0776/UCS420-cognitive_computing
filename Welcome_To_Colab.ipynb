```
n = int(input("Enter a no: "))
if 'sum' in locals() and isinstance(sum, int):
    del sum
is_prime = [True]*(n+1)
is_prime[0] = is_prime[1]=False
for i in range(2 , n//2+1):
  if is_prime[i]:
    for j in range(i*i ,n+1 , i):
      is_prime[j]= False
ans = sum(num for num , prime in enumerate(is_prime) if prime)
print(ans)
```

```
def IsPrime(n):
  for i in range(2 , n//2 + 1):
    if n%i==0:
      return 0
  return 1
print(IsPrime(20))
print(IsPrime(23))
print(IsPrime(200))
print(IsPrime(37))
```
