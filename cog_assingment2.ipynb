```python
#Q3
import random
random.seed(roll_no)
rands = [random.randint(100,900) for _ in range(100)]
odd_nums = [num for num in rands if num % 2 != 0]
print(odd_nums)
odd_count = sum(1 for x in rands if x % 2 != 0)
print(odd_count)
even_nums = [i for i in rands if i%2 ==0]
print(even_nums)
even_count = sum(1 for x in rands if x % 2 == 0)
print(even_count)
```

```python
def is_prime(n):
    if n <2:
        return False
    for i in range(2 , int(n**0.5)+1):
        if n% i ==0:
            return False
    return True

prime_numbs = [num for num in rands if is_prime(num)]
print(prime_numbs)
print(len(prime_numbs))
```

```python
most_frequent = max(rands , key = rands.count)
freq = rands.count(most_frequent)
print(most_frequent)
print(freq)
```
